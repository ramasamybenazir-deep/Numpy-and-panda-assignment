```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperature Array:")
print(temperatures)
```

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

```python
week1 = temperatures[0]

print("Week 1:")
print(week1)
```

```python
week2 = temperatures[1]

print("Week 2:")
print(week2)
```

```python
week1_weekend = temperatures[0, -2:]

print("Week 1 Weekend:")
print(week1_weekend)
```

```python
week2_weekend = temperatures[1, -2:]

print("Week 2 Weekend:")
print(week2_weekend)
```

```python
weekends = temperatures[:, -2:]

print("Weekends for Both Weeks:")
print(weekends)
```

```python
#  create Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Marks Series:")
print(marks)
```

```python
first_rank = marks.iloc[0]

print("Mark of 1st Rank:", first_rank)
```

```python
top_three = marks.loc[['Rank1', 'Rank2', 'Rank3']]

print("Top 3 Ranks:")
print(top_three)
```

```python
third_rank = marks.iloc[2]

print("Mark of 3rd Rank:", third_rank)
above_90 = marks[marks > 90]

print("Marks Greater Than 90:")
print(above_90)
```

```python
marks.loc['Rank1'] = 100

print("After modifying Rank 1:")
print(marks)
```

```python
marks = marks.drop('Rank5')

print("After removing Rank 5:")
print(marks)
```

```python
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
print(transactions)
```

```python
print(transactions.head())
```

```python
print(transactions.tail())
```

```python
print("Shape:", transactions.shape)
```

```python
print("Column Names:")
print(transactions.columns)
```

```python
print("Data Types:")
print(transactions.dtypes)
```

```python
transactions.info()
```

```python
print(transactions[['ProductCategory', 'Amount']])
```

```python
print(transactions.iloc[:, -3:])
```

```python
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_data)
```

```python
category_counts = transactions['ProductCategory'].value_counts()

print(category_counts)
```

```python
unique_regions = transactions['Region'].unique()

print("Unique Regions:")
print(unique_regions)
```

```python
mean_amount_region = transactions.groupby('Region')['Amount'].mean()

print("Mean Amount by Region:")
print(mean_amount_region)
```

```python
transactions.loc[
    transactions['TransactionID'] == 102,
    'Amount'
] = 165

print(transactions)
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

```python
transactions = transactions[
    transactions['TransactionID'] != 109
]

print(transactions)
```

```python
transactions = transactions.drop('Discount', axis=1)

print(transactions)
```
